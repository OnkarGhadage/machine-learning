```python
import pandas as pd
```

```python
df = pd.read_csv("breast-cancer.csv")
```

```python
df
```

```python
df.drop(columns='id', inplace=True)
```

```python
df
```

```python
# i am not a bio student but thing i get to know it that
# M is Malignant and B is Benign: both are tumers and M is cancerous and B is not.
# So M:1 and B:0
df["diagnosis"] = df["diagnosis"].apply(lambda x: 1 if x=='M' else 0)
```

```python
df
```

```python
df['diagnosis'].value_counts()
```

```python
df.isnull().sum().sum()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns="diagnosis"), df['diagnosis'], test_size=0.2, random_state=4)
```

```python
pipeline = Pipeline([
    ("preprocess", StandardScaler()),
    ("knn", KNeighborsClassifier())
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
y_test.value_counts()
```

```python
accuracy_score(y_test, y_pred)
```

```python
for i in range(1, 20):
    pipeline = Pipeline([
        ("preprocess", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=i))
    ])
    pipeline.fit(X_train, y_train)
    y_pred = pipeline.predict(X_test)
    print(i, accuracy_score(y_test, y_pred))
```

# Finding best value for n_features using GridSearchCV

```python
par_grid = {"knn__n_neighbors":list(range(1, 50))}
grid_search = GridSearchCV(pipeline, par_grid, cv=5, scoring='accuracy')
```

```python
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_estimator_
```
