```python
import pandas as pd
import numpy as np
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
np.random.seed(9) #to get same sample every time
df = df.sample(5)
```

```python
df.reset_index(drop=True, inplace=True)
df
```

```python
#Lets add missing values
df.iloc[0, 0] = np.nan
df.iloc[3, 1] = np.nan
df.iloc[4, 2] = np.nan
```

```python
X = df.iloc[:, :-1]
X
```

```python
y = df.iloc[:, 3:]
y
```

```python
ii = IterativeImputer()
```

```python
imputed = ii.fit_transform(X)
```

```python
df1 = pd.DataFrame(imputed, columns=X.columns)
```
