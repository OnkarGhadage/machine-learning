```python
import  numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('train.csv', usecols=['GarageQual','FireplaceQu','SalePrice'])
```

```python
df.isnull().mean()*100
```

```python
df['FireplaceQu'].value_counts()/df['FireplaceQu'].value_counts().sum()*100, df['GarageQual'].value_counts()/df['GarageQual'].value_counts().sum()*100
```

As we can see the `FireplaceQu` has the major part missing values, but the `GarageQual` have only 5.5% values missing. Which is okay to fill with both most frequent and 'missing'.  
So will just apply the both imputation on GarageQual. Cause it satisfies all the assumptions or requirements to fill the missing values.

## Most frequent imputation

```python
df['GarageQual_mf_py'] = df['GarageQual'].fillna(df['GarageQual'].mode()[0])
```

```python
df['GarageQual_mf_py'].isnull().sum()
```

```python
df['GarageQual'].value_counts(), df['GarageQual_mf_py'].value_counts()
```

### In this the value count of the most frequent vallues increases lil bit cause we are replacing all values with most frequent.

```python
mf = SimpleImputer(strategy='most_frequent')
```

```python
df['GarageQual_mf_sk'] = mf.fit_transform(df.iloc[:,1:2]).ravel()
```

```python
df['GarageQual_mf_sk'].isnull().sum()
```

## Missing category imputation

```python
df['GarageQual_mc_py'] = df['GarageQual'].fillna('Missing')
```

```python
df['GarageQual_mc_py'].isnull().sum()
```

```python
df['GarageQual_mc_py'].value_counts()
```

### This introduct the new categoory `Missing` in the feature

```python
mc = SimpleImputer(strategy='constant', fill_value='Missing')
```

```python
b = mc.fit_transform(df.loc[:,['GarageQual']])
```

```python
df['GarageQual_mc_sk'] = b.ravel()
```

```python
df['GarageQual_mc_sk'].value_counts()
```
