```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv('mnist_train.csv')
test = pd.read_csv('mnist_test.csv')
```

```python
df = pd.concat([train, test], axis=0)
```

```python
df
```

```python
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from mlxtend.plotting import plot_decision_regions
from sklearn.preprocessing import PolynomialFeatures
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns='label'), df['label'], test_size=0.2, random_state=42)
```

```python
pca = PCA(n_components=2)
```

```python
X_train_trf = pca.fit_transform(X_train)
X_test_trf = pca.transform(X_test)
```

```python
plt.scatter(x=X_train_trf[:,:1], y=X_train_trf[:,1:], c=y_train)
```

```python
lg = LogisticRegression()
```

```python
lg.fit(X_train_trf, y_train)
```

```python
plt.figure(figsize=(18,12))
plot_decision_regions(X_train_trf, y_train.values, clf=lg)
```

```python
poly = PolynomialFeatures(degree=2)
```

```python
X_train_trf_poly = poly.fit_transform(X_train_trf)
X_test_trf_poly = poly.transform(X_test_trf)
```

```python
lg.fit(X_train_trf_poly, y_train)
```

```python
y_train.values
```

```python
plt.figure(figsize=(18,12))
filler_values = {
    0: 1, # Bias term
    3: X_train_trf_poly[:, 3].mean(), # Mean of PCA1^2
    4: X_train_trf_poly[:, 4].mean(), # Mean of PCA1*PCA2
    5: X_train_trf_poly[:, 5].mean()  # Mean of PCA2^2
}
plot_decision_regions(X_train_trf_poly, y_train.values, clf=lg, feature_index=[1, 2], filler_feature_values=filler_values)
```
