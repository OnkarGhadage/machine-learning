```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
def perceptron(X, y):
    X = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1

    for _ in range(100):
        s = np.random.randint(0, 100)
        y_hat = 1 if np.dot(X[s], weights) > 0 else 0
        weights += lr * (y[s] - y_hat) * X[s]

    return weights[0], weights[1:]
```

```python
perceptron(X, y)
```

```python
class mylogic:
    def __init__(self, learning_rate, epochs):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.intercept_ = None
        self.coef_ = None
    def fit(self, X, y):
        X_ = np.insert(X, 0, 1, axis=1)
        weights = np.ones(X_.shape[1])
        for i in range(self.epochs):
            s = np.random.randint(0, 100)
            weights = weights + self.learning_rate * (y[s] - 1 if np.dot(weights, X_[s]) > 0 else 0) * X_[s]
        self.intercept_ = weights[0]
        self.coef_ = weights[1:]
    def predict(self, X):
        return 1 if (np.dot(X, self.coef_) + self.intercept_) > 0 else 0
```

```python
s1 = mylogic(learning_rate=0.1, epochs=1000)
```

```python
s1.fit(X, y)
```

```python
s1.predict(X[1])
```
