# Ordinal Encoding

```python
df.review.unique()
```

```python
df.education.unique()
```

```python
review_map = {
    'Poor' : 0,
    'Average' : 1,
    'Good': 2
}
edu_map = {
    'School' : 0,
    'UG' : 1,
    'PG': 2
}
```

```python
df['review-encoded'] = df['review'].map(review_map)
df['edu-encoded'] = df['education'].map(edu_map)
```
