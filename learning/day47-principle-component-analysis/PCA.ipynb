```python
import numpy as np
import pandas as pd

np.random.seed(23) 

mu_vec1 = np.array([0,0,0])
cov_mat1 = np.array([[1,0,0],[0,1,0],[0,0,1]])
class1_sample = np.random.multivariate_normal(mu_vec1, cov_mat1, 20)

df = pd.DataFrame(class1_sample,columns=['feature1','feature2','feature3'])
df['target'] = 1

mu_vec2 = np.array([1,1,1])
cov_mat2 = np.array([[1,0,0],[0,1,0],[0,0,1]])
class2_sample = np.random.multivariate_normal(mu_vec2, cov_mat2, 20)

df1 = pd.DataFrame(class2_sample,columns=['feature1','feature2','feature3'])

df1['target'] = 0

df = pd.concat([df, df1],ignore_index=True)

df = df.sample(40)
```

```python
df.head()
```

```python
# 1. Apply scaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
df.iloc[:, :3] = scaler.fit_transform(df.iloc[:, :3])
```

```python
df.head()
```

```python
# 2. Calculate covariance matrix
coMatrix = np.cov([df.iloc[:, 0], df.iloc[:, 1], df.iloc[:,2]])
```

```python
coMatrix
```

Diagonal elements are the variance of 0, 1 and 2nd columns.

```python
df['feature1'].var()
```

```python
# 3. Calculate eigenvectors and eigenvalues
evalues, evectors = np.linalg.eig(coMatrix)
```

```python
evalues, evectors
```

All the values are in complex form, `real + imaginary`.  
Use `.real` to get only real part i.e. pure values

```python
evalues.real
```

```python
evectors.real
```

### Lets convert the 3d space into 2d.
## That is we are converting the 3D data into 2D with highest possible variance wich will help to preserve the maximum iinformation.

```python
evectors = evectors[:2]
```

```python
evectors
```

```python
# 4. Transform the df.
new_df = np.dot(df.iloc[:, :3], evectors.T).real
```

```python
new_df = pd.DataFrame(new_df, columns=df.iloc[:,:2].columns)
```

```python
new_df['target'] = df['target'].values
```
