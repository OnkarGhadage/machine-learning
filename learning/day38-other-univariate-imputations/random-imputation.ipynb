### It is not safe to do random sample imputation on numeric data if missing values are >5% and for categorical data the safe side is <10%

lest just do on numerical data

```python
df['Age_imputed'] = df['Age']
```

```python
df.loc[df['Age_imputed'].isnull(), 'Age_imputed'] = df['Age'].dropna().sample(df['Age_imputed'].isnull().sum(), replace=True).values
```
