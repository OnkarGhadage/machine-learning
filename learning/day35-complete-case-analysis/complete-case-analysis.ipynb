```python
import pandas as pd
```

```python
df = pd.read_csv('data_science_job.csv')
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
drop_col = [col for col in df.columns if df[col].isnull().mean() < 0.05 and df[col].isnull().mean() > 0]
drop_col
```

```python
df.shape
```

```python
#lets drop from columns 
new = df.dropna(subset=drop_col)
```

```python
new.isnull().mean()*100
```

## Lets check whether the dropping the rows had any impact on the data.

### For numeric columns

```python
import matplotlib.pyplot as plt
import seaborn as sns
for i in df.describe().columns:
    sns.kdeplot(data=df, x=i)
    sns.kdeplot(data=new, x=i)
    plt.show()
```
